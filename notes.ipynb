### Pandas DataTypes
- series(s) : single dimensional array. Single column.
    - homogenous : 1 data type per series
- dataframe(df) : two dimen array (table)

```python
#declare series
ser = pd.Series(list('abcdefghijklmnopqrstuvwxyz')) # vertical columns 
arr = np.array(list('abcdefghijklmnopqrstuvwxyz'))
print(arr)

#label with name
ser.name = 'alphabet'

# .head(n) shows top 5 records by default - n is number of records
print(ser.head())

# .tail(n) shows the bottom 5 records 
print(ser.tail())
```

```python
# declare a series from a dictionary
d = {
    'a': 0.0,
    'b': 1.0,
    'c': 2.0
}
# for a dict becoming a series, the keys become the indexes
float_series = pd.Series(d)
print(float_series)
```

```python
#This pulls out the specified index and it's corresponding data
custom_index = pd.Series(d, index=['a', 'y', 'c'])
print(custom_index)
```

```python
calories = {
    'day1': 2200,
    'day2': 2500,
    'day3': 2700
}

index = ['day1', 'day2']

# create a series and pull out only days one and two
calories_series = pd.Series(calories, index=index)
print(calories_series)
```

```python
scalar_series = pd.Series(5.0, index=['a', 'b', 'c', 'd', 'e'])

print(scalar_series)
```

```python
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

# just like arrays math within series are performed element wise
print(ser1 + ser2)
print(ser1 - ser2)
print(ser1 * ser2)
print(ser1 / ser2)
```

```python
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

# can use comparison operators to compare elements in a series
print(ser1 < ser2) #false for each element
```

### Working w/DataFrames & JSON

```python
cars_df = pd.read_json('data/cars.json')

#look at the first and last 5 entries
print(cars_df.head())
```

```python
#print last 5
print(cars_df.tail())
```

```python
# look at the descriptive statistics
print(cars_df.describe())
```

```python
# looks at the shape
rows = cars_df.shape[0]
columns = cars_df.shape[1]

print('Rows/Records: ', rows)
print('Columns: ', columns)
```

```python
rows, cols = cars_df.shape
print('Rows: ', rows)
print('Cols: ', cols)
```

```python
# looks at the dtypes
print(cars_df.dtypes)
```

```python
# num of non null values
print(cars_df.count())
```

```python
# check for null or missing values
print(cars_df.isnull().sum())
```

```python
print(cars_df.info())
```

```python
# Value counts & values
# #print(cars_df['Car'].value_counts())
print(cars_df['city'].value_counts())
```

```python
#accessing  single values
# .at() label based
print(cars_df.at[143, 'Car'])

# .iat() index based
print(cars_df.iat[143, 0])
```

```python
# Scenario: you need to loop over ea column and perform an action on it
print(cars_df.columns)

#turn columns into iterable list
print(cars_df.columns.tolist())
```

```python
# Another way to find rows and cols, 0 = rows, 1 =columns
print(cars_df.axes)
```

```python
print(len(cars_df.axes[0]))
```
