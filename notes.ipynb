### Pandas DataTypes
- series(s) : single dimensional array. Single column.
    - homogenous : 1 data type per series
- dataframe(df) : two dimen array (table)

```python
#declare series
ser = pd.Series(list('abcdefghijklmnopqrstuvwxyz')) # vertical columns 
arr = np.array(list('abcdefghijklmnopqrstuvwxyz'))
print(arr)

#label with name
ser.name = 'alphabet'

# .head(n) shows top 5 records by default - n is number of records
print(ser.head())

# .tail(n) shows the bottom 5 records 
print(ser.tail())
```

```python
# declare a series from a dictionary
d = {
    'a': 0.0,
    'b': 1.0,
    'c': 2.0
}
# for a dict becoming a series, the keys become the indexes
float_series = pd.Series(d)
print(float_series)
```

```python
#This pulls out the specified index and it's corresponding data
custom_index = pd.Series(d, index=['a', 'y', 'c'])
print(custom_index)
```

```python
calories = {
    'day1': 2200,
    'day2': 2500,
    'day3': 2700
}

calories['day4'] = 3000 # adding values to a dictionary

index = ['day1', 'day2']

# create a series and pull out only days one and two
calories_series = pd.Series(calories, index=index)

print(calories_series)

day_index_info = calories_series.index.tolist()
print(day_index_info)
```

```python
scalar_series = pd.Series(5.0, index=['a', 'b', 'c', 'd', 'e'])

print(scalar_series)
```

```python
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

# just like arrays math within series are performed element wise
print(ser1 + ser2)
print(ser1 - ser2)
print(ser1 * ser2)
print(ser1 / ser2)
```

```python
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

# can use comparison operators to compare elements in a series
print(ser1 < ser2) #false for each element
```

# Slicing a DataFrame
- pull out specific parts of a dataframe
- examine a subset
- remove unwanted/unneeded data
- data isolation, remove clutter
- `df[[col1, col2, col3, col4]]` : slicing using double square bracket syntax

```python
# Initializing the nested list with Data set
employee_list = [['James', 36, 75, 5428000],
               ['Villers', 38, 74, 3428000],
               ['VKole', 31, 70, 8428000],
               ['Smith', 34, 80, 4428000],
               ['Gayle', 40, 100, 4528000],
               ['Rooter', 33, 72, 7028000],
               ['Peterson', 42, 85, 2528000],
               ['John', 41, 85, 1528000],

]

columns = ['Name', 'Age', 'Weight (kgs)', 'Salary']

df = pd.DataFrame(employee_list, columns=columns)
print(df.head())
```

# Slicing Functions
- `df.loc[]` - label based location (column names to locate and slice)
- `df.iloc[start:stop:step (rows), start:stop:step (cols)]` - index based location (index numbers to locate and slice)
- stop is always non-inclusive

```python
#python indexing operator
emp_subset = df.iloc[0:2, 0:2]
print(emp_subset)
```
